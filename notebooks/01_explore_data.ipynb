```python
import pandas as pd
import numpy as np

file = "../data/raw/02-20-2018.csv"

bad_summary = {}

for i, chunk in enumerate(
    pd.read_csv(file, low_memory=False, chunksize=200000)
):
    # Remove identity columns exactly like the cleaner
    id_cols = ['Flow ID', 'Src IP', 'Src Port', 'Dst IP']
    chunk = chunk.drop(columns=[c for c in id_cols if c in chunk.columns])

    numeric_cols = [
        c for c in chunk.columns
        if c not in ['Timestamp', 'Label']
    ]

    for col in numeric_cols:
        converted = pd.to_numeric(chunk[col], errors='coerce')
        bad = converted.isna().sum()

        if bad > 0:
            bad_summary[col] = bad_summary.get(col, 0) + bad

    if (i + 1) % 5 == 0:
        print(f"Checked chunk {i + 1}")

print("\n=== BAD VALUES SUMMARY ===")

if bad_summary:
    for col, count in sorted(bad_summary.items(), key=lambda x: -x[1]):
        print(f"{col}: {count}")
else:
    print("No non-numeric values found.")
```

```python
import pandas as pd
import os

processed_dir = "../data/processed"
files = sorted([f for f in os.listdir(processed_dir) if f.startswith("cleaned_") and f.endswith(".csv")])

# CHECK 1: constant columns across all files
constant_per_file = {}
for f in files:
    df = pd.read_csv(os.path.join(processed_dir, f))
    constant_per_file[f] = set([col for col in df.columns if df[col].nunique() <= 1])
    del df

always_constant = set.intersection(*constant_per_file.values())
sometimes_constant = set.union(*constant_per_file.values()) - always_constant
print("CHECK 1 — Always constant (safe to drop):", always_constant)
print("CHECK 1 — Sometimes constant (keep):", sometimes_constant)

# CHECK 2: overall label counts combined across all files
label_totals = {}
for f in files:
    df = pd.read_csv(os.path.join(processed_dir, f))
    counts = df['Label'].value_counts().to_dict()
    for label, count in counts.items():
        label_totals[label] = label_totals.get(label, 0) + count
    del df

print("\nCHECK 2 — Combined label totals across all 10 files:")
for label, count in sorted(label_totals.items(), key=lambda x: -x[1]):
    print(f"  {label}: {count}")

# CHECK 3: timestamp range per file (confirms chronological order, needed for windowing)
print("\nCHECK 3 — Timestamp range per file:")
for f in files:
    df = pd.read_csv(os.path.join(processed_dir, f), usecols=['Timestamp'], parse_dates=['Timestamp'])
    print(f"  {f}: {df['Timestamp'].min()} to {df['Timestamp'].max()}")
    del df
```

```python
import pandas as pd
import numpy as np

# 1. Load ONE raw CSV file (replace with your actual file path)
file_path = "path/to/your/cic_ids_2018_day_x.csv"
print(f"Loading data from {file_path}...")
df = pd.read_csv(file_path)

# Ensure Timestamp is a datetime object for windowing
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# 2. Check raw flow-level class distribution
print("\n--- Raw Flow-Level Class Distribution ---")
print(df['Label'].value_counts())

# 3. Simulate 30-second windowing to test Fix 2
# Calculate the percentage of each label within every 30-second window
windowed = df.groupby(pd.Grouper(key='Timestamp', freq='30s'))['Label'].value_counts(normalize=True).unstack(fill_value=0)

# 4. Apply the Old Rule (Majority > 50%)
def get_majority_label(row):
    top_label = row.idxmax()
    return top_label if row[top_label] > 0.50 else 'Benign'

old_rule_labels = windowed.apply(get_majority_label, axis=1)

# 5. Apply the New Rule (Threshold > 5%)
def get_threshold_label(row):
    # Drop Benign to focus on attack thresholds
    attack_cols = [col for col in row.index if col != 'Benign']
    if not attack_cols:
        return 'Benign'
    
    # Find the attack with the highest percentage
    attacks = row[attack_cols]
    top_attack = attacks.idxmax()
    
    # If the top attack is > 5%, label it as that attack
    if attacks[top_attack] > 0.05:
        return top_attack
    return 'Benign'

new_rule_labels = windowed.apply(get_threshold_label, axis=1)

# 6. Compare the Results
print("\n--- Window Labels: OLD RULE (>50% Majority) ---")
print(old_rule_labels.value_counts())

print("\n--- Window Labels: NEW RULE (>5% Threshold) ---")
print(new_rule_labels.value_counts())
```

```python
# ============================================================
# CHRONEX DATASET AUDIT
# READ-ONLY DIAGNOSTIC
# ============================================================

import os
import json
import numpy as np
import pandas as pd

print("=" * 80)
print("CHRONEX DATASET AUDIT")
print("=" * 80)


# ------------------------------------------------------------
# 1. ENVIRONMENT
# ------------------------------------------------------------

print("\n[1] ENVIRONMENT")
print("-" * 80)

print("Working directory:")
print(os.getcwd())

print("\nFiles in current directory:")
for f in os.listdir("."):
    print("  ", f)


# ------------------------------------------------------------
# 2. FIND NPZ / JSON / JOBLIB FILES
# ------------------------------------------------------------

print("\n[2] AVAILABLE DATA ARTIFACTS")
print("-" * 80)

extensions = [".npz", ".json", ".joblib"]

for root, dirs, files in os.walk("."):
    for f in files:
        if any(f.lower().endswith(ext) for ext in extensions):
            print(os.path.join(root, f))


# ------------------------------------------------------------
# 3. LOCATE DATASET
# ------------------------------------------------------------

print("\n[3] DATASET")
print("-" * 80)

possible_npz = [
    "./data/processed/model_ready/dataset.npz",
    "./dataset.npz",
    "dataset.npz"
]

dataset_path = None

for p in possible_npz:
    if os.path.exists(p):
        dataset_path = p
        break

if dataset_path is None:
    # Search recursively
    for root, dirs, files in os.walk("."):
        for f in files:
            if f == "dataset.npz":
                dataset_path = os.path.join(root, f)
                break
        if dataset_path:
            break

if dataset_path is None:
    print("❌ dataset.npz NOT FOUND")
else:
    print("Dataset:", dataset_path)

    data = np.load(dataset_path, allow_pickle=True)

    print("\nKeys:")
    for key in data.files:
        print("  ", key)

    for key in data.files:
        arr = data[key]
        print(f"\n{key}:")
        print("  shape:", arr.shape)
        print("  dtype:", arr.dtype)

        if arr.size > 0:
            print("  min:", np.nanmin(arr))
            print("  max:", np.nanmax(arr))
            print("  mean:", np.nanmean(arr))
            print("  std:", np.nanstd(arr))


# ------------------------------------------------------------
# 4. LOAD X/y
# ------------------------------------------------------------

print("\n[4] TRAIN / VALIDATION / TEST")
print("-" * 80)

if dataset_path:

    X_train = data["X_train"]
    X_val   = data["X_val"]
    X_test  = data["X_test"]

    y_train = data["y_train"]
    y_val   = data["y_val"]
    y_test  = data["y_test"]

    for name, arr in [
        ("X_train", X_train),
        ("X_val", X_val),
        ("X_test", X_test),
        ("y_train", y_train),
        ("y_val", y_val),
        ("y_test", y_test)
    ]:
        print(f"{name}: shape={arr.shape}, dtype={arr.dtype}")


# ------------------------------------------------------------
# 5. INPUT DIMENSIONS
# ------------------------------------------------------------

print("\n[5] MODEL INPUT DIMENSIONS")
print("-" * 80)

if dataset_path:

    print("Sequence length:", X_train.shape[1])
    print("Feature count:", X_train.shape[2])

    print("\nExpected CHRONEX:")
    print("  Sequence length = 10")
    print("  Features        = 71")

    print("\nChecks:")
    print("  Sequence length correct:",
          X_train.shape[1] == 10)

    print("  Feature count correct:",
          X_train.shape[2] == 71)


# ------------------------------------------------------------
# 6. LABEL DISTRIBUTION
# ------------------------------------------------------------

print("\n[6] LABEL DISTRIBUTION")
print("-" * 80)

def print_distribution(name, y):
    values, counts = np.unique(y, return_counts=True)

    print(f"\n{name}")
    print("-" * 40)

    total = len(y)

    for v, c in zip(values, counts):
        print(
            f"{str(v):30s} "
            f"{c:8d} "
            f"{c/total*100:8.3f}%"
        )

    print("Total:", total)


if dataset_path:
    print_distribution("TRAIN", y_train)
    print_distribution("VALIDATION", y_val)
    print_distribution("TEST", y_test)


# ------------------------------------------------------------
# 7. CLASS PRESENCE ACROSS SPLITS
# ------------------------------------------------------------

print("\n[7] CLASS PRESENCE ACROSS SPLITS")
print("-" * 80)

if dataset_path:

    train_classes = set(np.unique(y_train))
    val_classes   = set(np.unique(y_val))
    test_classes  = set(np.unique(y_test))

    all_classes = sorted(
        train_classes |
        val_classes |
        test_classes
    )

    print(
        f"{'CLASS':30s}"
        f"{'TRAIN':>10s}"
        f"{'VAL':>10s}"
        f"{'TEST':>10s}"
    )

    print("-" * 65)

    for cls in all_classes:
        print(
            f"{str(cls):30s}"
            f"{'YES' if cls in train_classes else 'NO':>10s}"
            f"{'YES' if cls in val_classes else 'NO':>10s}"
            f"{'YES' if cls in test_classes else 'NO':>10s}"
        )

    print("\nMissing from TRAIN:",
          sorted(all_classes - train_classes))

    print("Missing from VAL:",
          sorted(all_classes - val_classes))

    print("Missing from TEST:",
          sorted(all_classes - test_classes))


# ------------------------------------------------------------
# 8. LABEL ENCODER
# ------------------------------------------------------------

print("\n[8] LABEL ENCODER")
print("-" * 80)

label_encoder_path = "./data/processed/model_ready/label_encoder.joblib"

if os.path.exists(label_encoder_path):

    try:
        import joblib

        le = joblib.load(label_encoder_path)

        print("Encoder:", label_encoder_path)
        print("Classes:")
        for i, cls in enumerate(le.classes_):
            print(f"  {i:2d} -> {cls}")

        print("\nNumber of encoded classes:", len(le.classes_))

    except Exception as e:
        print("Could not load label encoder:")
        print(e)

else:
    print("label_encoder.joblib not found")


# ------------------------------------------------------------
# 9. METADATA
# ------------------------------------------------------------

print("\n[9] METADATA")
print("-" * 80)

metadata_path = "./data/processed/model_ready/metadata.json"

if os.path.exists(metadata_path):

    with open(metadata_path, "r", encoding="utf-8") as f:
        metadata = json.load(f)

    print(json.dumps(metadata, indent=2))

else:
    print("metadata.json not found")


# ------------------------------------------------------------
# 10. CHECK FOR NaN / INF
# ------------------------------------------------------------

print("\n[10] NaN / INF CHECK")
print("-" * 80)

if dataset_path:

    for name, arr in [
        ("X_train", X_train),
        ("X_val", X_val),
        ("X_test", X_test)
    ]:

        print(
            f"{name}: "
            f"NaN={np.isnan(arr).sum():,}, "
            f"Inf={np.isinf(arr).sum():,}"
        )


# ------------------------------------------------------------
# 11. CHECK SCALE
# ------------------------------------------------------------

print("\n[11] SCALE CHECK")
print("-" * 80)

if dataset_path:

    for name, X in [
        ("TRAIN", X_train),
        ("VAL", X_val),
        ("TEST", X_test)
    ]:

        flat = X.reshape(-1, X.shape[-1])

        print(f"\n{name}")

        print("  Overall mean:", np.mean(flat))
        print("  Overall std :", np.std(flat))

        print("  Feature means:")
        print("   min =", np.min(np.mean(flat, axis=0)))
        print("   max =", np.max(np.mean(flat, axis=0)))

        print("  Feature stds:")
        print("   min =", np.min(np.std(flat, axis=0)))
        print("   max =", np.max(np.std(flat, axis=0)))


# ------------------------------------------------------------
# 12. LOAD SCALER
# ------------------------------------------------------------

print("\n[12] SCALER")
print("-" * 80)

scaler_path = "./data/processed/model_ready/scaler.joblib"

if os.path.exists(scaler_path):

    try:
        import joblib

        scaler = joblib.load(scaler_path)

        print("Scaler:", scaler_path)
        print("Type:", type(scaler).__name__)

        if hasattr(scaler, "n_features_in_"):
            print("n_features_in_:", scaler.n_features_in_)

        if hasattr(scaler, "mean_"):
            print("Number of scaler means:", len(scaler.mean_))

            print("\nFirst 10 scaler means:")
            print(scaler.mean_[:10])

        if hasattr(scaler, "scale_"):
            print("\nFirst 10 scaler std/scales:")
            print(scaler.scale_[:10])

    except Exception as e:
        print("Could not load scaler:")
        print(e)

else:
    print("scaler.joblib not found")


# ------------------------------------------------------------
# 13. FEATURE NAMES
# ------------------------------------------------------------

print("\n[13] FEATURE NAMES")
print("-" * 80)

feature_names = None

if "metadata" in locals():

    possible_keys = [
        "feature_columns",
        "feature_columns_order",
        "features",
        "model_features"
    ]

    for key in possible_keys:
        if key in metadata:
            feature_names = metadata[key]
            print("Found metadata key:", key)
            break

if feature_names is not None:

    print("Number of features:", len(feature_names))

    for i, feature in enumerate(feature_names):
        print(f"{i+1:3d}. {feature}")

else:
    print("Feature names not found in expected metadata keys.")


# ------------------------------------------------------------
# 14. ATTACK_RATIO CHECK
# ------------------------------------------------------------

print("\n[14] ATTACK_RATIO CHECK")
print("-" * 80)

if feature_names is not None:

    attack_ratio_present = any(
        "attack_ratio" in str(x).lower()
        for x in feature_names
    )

    print("attack_ratio present in model features:",
          attack_ratio_present)

    if attack_ratio_present:
        print("⚠️ WARNING: attack_ratio is still in model input.")

    else:
        print("✅ attack_ratio is NOT in model input.")


# ------------------------------------------------------------
# 15. FLOW_COUNT CHECK
# ------------------------------------------------------------

print("\n[15] FLOW_COUNT CHECK")
print("-" * 80)

if feature_names is not None:

    flow_count_present = any(
        str(x).lower() == "flow_count"
        for x in feature_names
    )

    print("flow_count present:", flow_count_present)


# ------------------------------------------------------------
# 16. TARGET / FEATURE LEAKAGE NAMES
# ------------------------------------------------------------

print("\n[16] POSSIBLE TARGET-LEAKAGE FEATURES")
print("-" * 80)

if feature_names is not None:

    suspicious_terms = [
        "label",
        "attack",
        "target",
        "class",
        "malicious",
        "benign"
    ]

    suspicious = []

    for feature in feature_names:

        name = str(feature).lower()

        if any(term in name for term in suspicious_terms):
            suspicious.append(feature)

    if suspicious:
        print("Potentially suspicious feature names:")
        for x in suspicious:
            print("  ⚠️", x)
    else:
        print("No obviously label-derived feature names detected.")


# ------------------------------------------------------------
# 17. SEQUENCE OVERLAP CHECK
# ------------------------------------------------------------

print("\n[17] SEQUENCE OVERLAP CHECK")
print("-" * 80)

print("""
IMPORTANT:
The arrays alone may not contain timestamps/window IDs.

If sequences were created like:

W1 W2 W3 ... W10 -> W11
W2 W3 W4 ... W11 -> W12
W3 W4 W5 ... W12 -> W13

then consecutive samples overlap heavily.

This cell cannot prove cross-split temporal leakage unless
window/day/time identifiers were preserved.

We therefore need to inspect the sequence-generation code
or preserved sequence metadata for a definitive answer.
""")


# ------------------------------------------------------------
# 18. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("AUDIT SUMMARY")
print("=" * 80)

if dataset_path:

    print("Dataset found             : ✅")
    print("Sequence length           :", X_train.shape[1])
    print("Feature count             :", X_train.shape[2])
    print("Train sequences           :", len(X_train))
    print("Validation sequences      :", len(X_val))
    print("Test sequences            :", len(X_test))
    print("Total sequences           :", len(X_train) + len(X_val) + len(X_test))

    print("\nClasses:")
    print("  Train:", len(np.unique(y_train)))
    print("  Val  :", len(np.unique(y_val)))
    print("  Test :", len(np.unique(y_test)))

    if feature_names is not None:
        print("\nFeature count from metadata:", len(feature_names))

print("\n" + "=" * 80)
print("END OF CHRONEX DATASET AUDIT")
print("=" * 80)
```

```python
import numpy as np
import joblib
from sklearn.preprocessing import StandardScaler

BASE = "../data/processed/model_ready"

data = np.load(f"{BASE}/dataset.npz")
old_scaler = joblib.load(f"{BASE}/scaler.joblib")

# 1. Pool existing scaled data
X_all_scaled = np.concatenate(
    [data["X_train"], data["X_val"], data["X_test"]],
    axis=0
)

y_all = np.concatenate(
    [data["y_train"], data["y_val"], data["y_test"]],
    axis=0
)

# 2. Recover raw/original feature values
n_features = X_all_scaled.shape[-1]

X_all_raw = old_scaler.inverse_transform(
    X_all_scaled.reshape(-1, n_features)
).reshape(X_all_scaled.shape)

# 3. Use the SAME verified stratified split
from sklearn.model_selection import train_test_split

X_train_raw, X_temp_raw, y_train, y_temp = train_test_split(
    X_all_raw,
    y_all,
    test_size=0.30,
    stratify=y_all,
    random_state=42
)

X_val_raw, X_test_raw, y_val, y_test = train_test_split(
    X_temp_raw,
    y_temp,
    test_size=0.50,
    stratify=y_temp,
    random_state=42
)

# 4. Fit NEW scaler ONLY on training data
new_scaler = StandardScaler()

new_scaler.fit(
    X_train_raw.reshape(-1, n_features)
)

# 5. Transform all splits using that same scaler
X_train_new = new_scaler.transform(
    X_train_raw.reshape(-1, n_features)
).reshape(X_train_raw.shape).astype(np.float32)

X_val_new = new_scaler.transform(
    X_val_raw.reshape(-1, n_features)
).reshape(X_val_raw.shape).astype(np.float32)

X_test_new = new_scaler.transform(
    X_test_raw.reshape(-1, n_features)
).reshape(X_test_raw.shape).astype(np.float32)

# 6. Checks
print("Shapes:")
print("Train:", X_train_new.shape, y_train.shape)
print("Val:  ", X_val_new.shape, y_val.shape)
print("Test: ", X_test_new.shape, y_test.shape)

print("\nFinite check:")
print("Train:", np.isfinite(X_train_new).all())
print("Val:  ", np.isfinite(X_val_new).all())
print("Test: ", np.isfinite(X_test_new).all())

print("\nScaler:")
print("Features:", new_scaler.n_features_in_)
print("First 5 new means:", new_scaler.mean_[:5])
print("First 5 new scales:", new_scaler.scale_[:5])
```
